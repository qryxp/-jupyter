# 04 · 业务看板（一键多图组合）

把接单最常用的几张图固化成一条命令直接套；也能临时拼任意组合。
返回 `dict`（Jupyter 里多张图依次渲染）或 `组合=True` 拼成单张看板。

```python
import da, pandas as pd
df = pd.read_csv("data/sample_region_sales.csv")
```

## 固化看板：6 个覆盖四域

```python
# 地区销售：柱状(地区×销售额) + 饼(占比) + 折线(月份趋势)
da.图表.看板(df, 名称="地区销售")
da.图表.看板清单()                      # 列出所有固化看板及其所需列
```

```python
# 其余 5 个固化看板
da.图表.看板(df, 名称="渠道对比")
da.图表.看板(df, 名称="收益趋势")
da.图表.看板(df, 名称="品类销售")
da.图表.看板(df, 名称="账号表现")
da.图表.看板(df, 名称="接单效率")
```

## 自定义看板：临时拼任意组合（不写函数）

```python
da.图表.自定义看板(df, 图表=[
    ("柱状图", "地区", "销售额"),
    ("饼图",   "地区", "销售额"),
    ("折线图", "月份", "销售额"),
])
```

## 全局设一次默认风格

```python
da.图表.设定(默认引擎="plotly", 默认预设="商务蓝")
da.图表.看板(df, 名称="地区销售")     # 直接出商务蓝
# 换黑底：预设="暗夜"；组合成单张：组合=True
```

## 缺列自动跳过

数据里没有某子图需要的列，看板会跳过该子图并提示，不让整块崩。
也可 `da.图表.查看配方(da.图表.地区销售看板)` 看拼装源码。
