全部用**中文调用**，像 `da.技能外包.净到手率(df)` 一样直接调。

```python
import da
import pandas as pd
print('da 已加载，业务域：', ['技能外包','小微电商','自媒体','零工平台'])
```

## 一、先清洗（多格式日期 + 伪空值自动填充）

示例数据里混了 `2026/7/3`、`16/07/2026`、`2026年7月5日` 三种日期，还有 `无`/`空` 这种伪空值。

```python
原始 = pd.read_csv('data/sample_skills.csv')
print('清洗前 日期 dtype：', 原始['日期'].dtype)
print(原始[['日期','评价分']].head())

干净 = da.数据清洗.清洗管道(原始)
print('\n清洗后 日期 dtype：', 干净['日期'].dtype)
print(干净[['日期','评价分']].head())
```

## 二、像调 pandas 一样算指标（域A 技能外包）

```python
df = da.数据清洗.清洗管道(pd.read_csv('data/sample_skills.csv'))
print('净到手率 :', round(da.技能外包.净到手率(df), 3))
print('客单价   :', round(da.技能外包.客单价(df), 2))
print('时薪     :', round(da.技能外包.时薪(df), 2))
print('返工率   :', round(da.技能外包.返工率(df), 3))
print('毛利率   :', round(da.技能外包.毛利率(df), 3))
```

## 五、其他域随手算

```python
ec = pd.read_csv('data/sample_ecommerce.csv')
print('销售额       :', da.小微电商.销售额(ec))
print('净销售额     :', da.小微电商.净销售额(ec))
print('客单价ATV    :', round(da.小微电商.客单价ATV(ec), 2))
print('销售毛利率   :', round(da.小微电商.销售毛利率(ec), 3))
print('整体转化率   :', round(da.小微电商.整体转化率(ec), 4))
print('动销率       :', round(da.小微电商.动销率(ec), 3))

md = pd.read_csv('data/sample_media.csv')
print('互动率       :', round(da.自媒体.互动率(md), 4))
print('粉丝净值     :', round(da.自媒体.粉丝净值(md), 4))

gg = pd.read_csv('data/sample_gig.csv')
print('中标率       :', round(da.零工平台.中标率(gg), 3))
print('纠纷率       :', round(da.零工平台.纠纷率(gg), 3))
```
