# 02 · 图表上手（Seaborn / Plotly / PyEcharts）

`da.图表` 让你用一句中文命令，直接把数据画成调好参数的图。
- 默认 `engine="plotly"`（交互、可 hover）
- `engine="seaborn"`（静态、适合论文/报告）
- `engine="pyecharts"`（HTML、适合网页/中文看板）

数据会自动按类别聚合，不用你先 groupby。

```python
import da
import pandas as pd

df = pd.read_csv("data/sample_region_sales.csv")
df.head()
```

## 1. 各地区的销售额柱状图（你举的例子）

直接 `x="地区", y="销售额"`，库会自动按地区求和并画好。

```python
# 交互式（默认）
da.图表.柱状图(df, x="地区", y="销售额", title="各地区销售额")
```

```python
# 静态出版风
da.图表.柱状图(df, x="地区", y="销售额", engine="seaborn", title="各地区销售额")
```

```python
# HTML 看板风
chart = da.图表.柱状图(df, x="地区", y="销售额", engine="pyecharts", title="各地区销售额")
chart
```

## 2. 堆叠 / 折线 / 饼图 / 热力图

```python
# 各地区各品类的堆叠柱状图
da.图表.堆叠柱状图(df, x="地区", y="销售额", stack="品类", title="各地区品类构成")
```

```python
# 各地区销售额随时间（月份）的折线图，按地区分多条线
da.图表.折线图(df, x="月份", y="销售额", color="地区", title="各月销售额走势")
```

```python
# 地区销售额占比（环形图）
da.图表.环形图(df, x="地区", y="销售额", title="地区销售占比")
```

```python
# 地区 × 月份 的销售额热力图
da.图表.热力图(df, x="月份", y="地区", z="销售额", title="地区×月份热力")
```

## 2.5 极简入口 `画` 与样式预设

不想记函数名？用 `画` + 中文类型名；预设填名字即可切换配色风格（清爽白/商务蓝/活力橙/暗夜）。

```python
# 极简入口：只给类型名，engine/预设 取默认（plotly + 清爽白）
da.图表.画(df, 类型="柱状图", x="地区", y="销售额")

# 预设：一行切换商务蓝风格
da.图表.画(df, 类型="柱状图", x="地区", y="销售额", 预设="商务蓝")

# 配置一次默认，之后连 engine/预设 都省
da.图表.设定(默认引擎="plotly", 默认预设="清爽白")
da.图表.预设示例()
```
